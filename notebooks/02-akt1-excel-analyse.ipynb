```python
import pandas as pd
from pathlib import Path

excel_path = Path("../Lets Meet DB Dump.xlsx")

print("Datei vorhanden:", excel_path.exists())

df = pd.read_excel(
    excel_path,
    sheet_name="Tabelle1",
    dtype=str,
    keep_default_na=False
)

print("Zeilen:", len(df))
print("Spalten:", len(df.columns))
```

```python
for nummer, spalte in enumerate(df.columns, start=1):
    print(nummer, repr(spalte))
```

```python
print("Leere Werte pro Spalte:")

for spalte in df.columns:
    anzahl = (df[spalte] == "").sum()
    print(f"{spalte}: {anzahl}")
```

```python
email_spalte = "E-Mail"

leere_emails = (df[email_spalte] == "").sum()

doppelte_emails = df.loc[
    df[email_spalte] != "",
    email_spalte
].duplicated().sum()

print("Leere E-Mails:", leere_emails)
print("Doppelte E-Mails:", doppelte_emails)
print("Eindeutige E-Mails:", df[email_spalte].nunique())
```

```python
hobby_spalte = df.columns[3]

df.loc[
    df[hobby_spalte] == "",
    ["Nachname, Vorname", "E-Mail", hobby_spalte]
]
```

```python
name_spalte = "Nachname, Vorname"

print(
    df[name_spalte]
    .str.count(", ")
    .value_counts()
    .sort_index()
)
```

```python
namen = df[name_spalte].str.split(", ", n=1, expand=True)

namen.columns = ["last_name", "first_name"]

display(namen.head(10))
```

```python
namen_mit_zusatzleerzeichen = (
    (namen["last_name"] != namen["last_name"].str.strip()) |
    (namen["first_name"] != namen["first_name"].str.strip())
)

print(
    "Namen mit zusätzlichen äußeren Leerzeichen:",
    namen_mit_zusatzleerzeichen.sum()
)

display(
    pd.concat(
        [
            df.loc[namen_mit_zusatzleerzeichen, [name_spalte]],
            namen.loc[namen_mit_zusatzleerzeichen]
        ],
        axis=1
    ).head(20)
)
```

```python
adresse_spalte = "Straße Nr, PLZ Ort"

print(
    df[adresse_spalte]
    .str.count(", ")
    .value_counts()
    .sort_index()
)
```

```python
display(
    df.loc[
        df[adresse_spalte].str.count(", ") > 2,
        ["Nachname, Vorname", adresse_spalte]
    ]
)
```

```python
adressen = df[adresse_spalte].str.split(", ", n=2, expand=True)

adressen.columns = ["street", "postal_code", "city"]

display(adressen.head(10))
```

```python
geburtsdatum_text = df["Geburtsdatum"]

geburtsdatum = pd.to_datetime(
    geburtsdatum_text,
    format="%d.%m.%Y",
    errors="coerce"
)

print("Ungültige Geburtsdaten:", geburtsdatum.isna().sum())
print("Frühestes Geburtsdatum:", geburtsdatum.min())
print("Spätestes Geburtsdatum:", geburtsdatum.max())
```

```python
migration_users = pd.DataFrame({
    "email": df["E-Mail"],
    "first_name": namen["first_name"],
    "last_name": namen["last_name"],
    "birth_date": geburtsdatum.dt.date,
    "postal_code": adressen["postal_code"],
    "city": adressen["city"]
})

display(migration_users.head(10))
```

```python
print("Zeilen:", len(migration_users))
print("Leere E-Mails:", (migration_users["email"] == "").sum())
print("Doppelte E-Mails:", migration_users["email"].duplicated().sum())
print("Ungültige Geburtsdaten:", migration_users["birth_date"].isna().sum())
print("Leere Postleitzahlen:", (migration_users["postal_code"] == "").sum())
print("Leere Städte:", (migration_users["city"] == "").sum())
```
